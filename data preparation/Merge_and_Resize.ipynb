```python
import os
import re
from collections import Counter
from google.colab import drive

# 1. Kết nối với Google Drive
drive.mount('/content/drive')

# 2. CẤU HÌNH ĐƯỜNG DẪN (Thay đổi đường dẫn đến thư mục ảnh của bạn)
# Ví dụ: '/content/drive/MyDrive/HinhAnh/BoAnh1'
folder_path = '/content/drive/MyDrive/data/images'

# Cấu hình dải số
START = 1
END = 8000

def check_images(path):
    if not os.path.exists(path):
        print("Đường dẫn không tồn tại! Vui lòng kiểm tra lại.")
        return

    all_files = os.listdir(path)

    existing_numbers = []
    duplicates = []
    out_of_range = []
    invalid_files = []

    # Regex để tìm số trong tên file (ví dụ: 00001.jpg hoặc 00001 (1).jpg)
    pattern = re.compile(r'(\d+)')

    for filename in all_files:
        # Bỏ qua các file hệ thống ẩn
        if filename.startswith('.'): continue

        match = pattern.search(filename)
        if match:
            num_val = int(match.group(1))
            existing_numbers.append(num_val)

            # Kiểm tra xem có nằm ngoài khoảng 1-8000 không
            if num_val < START or num_val > END:
                out_of_range.append(filename)
        else:
            # File không có số trong tên
            invalid_files.append(filename)

    # 1. Tìm ảnh thiếu
    missing = [i for i in range(START, END + 1) if i not in existing_numbers]

    # 2. Tìm ảnh trùng (xuất hiện nhiều lần)
    count_map = Counter(existing_numbers)
    duplicates_list = [num for num, count in count_map.items() if count > 1]

    # In kết quả
    print(f"--- BÁO CÁO KIỂM TRA (Từ {START:05d} đến {END:05d}) ---")
    print(f"Tổng số file tìm thấy: {len(all_files)}")

    print(f"\n1. ❌ THIẾU: {len(missing)} ảnh")
    if missing:
        if len(missing) > 20:
            print(f"   Danh sách (20 số đầu): {missing[:20]} ...")
        else:
            print(f"   Danh sách: {missing}")

    print(f"\n2. 📂 TRÙNG LẶP: {len(duplicates_list)} số bị lặp lại")
    if duplicates_list:
        for num in duplicates_list:
            # Tìm các tên file thực tế liên quan đến số bị trùng
            related_files = [f for f in all_files if pattern.search(f) and int(pattern.search(f).group(1)) == num]
            print(f"   Số {num:05d} xuất hiện ở các file: {related_files}")

    print(f"\n3. ⚠️ DƯ RA (Ngoài khoảng {START}-{END}): {len(out_of_range)} file")
    if out_of_range:
        print(f"   Danh sách: {out_of_range[:10]} {'...' if len(out_of_range) > 10 else ''}")

    print(f"\n4. 🚫 FILE KHÔNG HỢP LỆ (Không chứa số): {len(invalid_files)} file")
    if invalid_files:
        print(f"   Danh sách: {invalid_files}")

# Chạy hàm kiểm tra
check_images(folder_path)
```

```python
import os
import shutil
from google.colab import drive

# 1. Kết nối Google Drive
drive.mount('/content/drive')

# 2. Cấu hình đường dẫn
folder_1 = '/content/drive/MyDrive/data/House/'       # Đường dẫn folder 1
folder_2 = '/content/drive/MyDrive/data/Transport/'       # Đường dẫn folder 2
folder_3 = '/content/drive/MyDrive/data/Street/'
target_folder = '/content/drive/MyDrive/8000 ảnh' # Folder đích sẽ chứa tất cả ảnh

# Tạo folder đích nếu chưa có
if not os.path.exists(target_folder):
    os.makedirs(target_folder)
    print(f"Đã tạo folder mới: {target_folder}")

def merge_folders(source_folders, destination):
    valid_exts = ('.jpg', '.jpeg', '.png', '.bmp', '.gif')
    total_copied = 0

    for src in source_folders:
        print(f"\nĐang lấy ảnh từ: {src}...")

        # Lấy danh sách file trong folder nguồn
        files = [f for f in os.listdir(src) if f.lower().endswith(valid_exts)]

        for filename in files:
            source_path = os.path.join(src, filename)
            dest_path = os.path.join(destination, filename)

            # Xử lý nếu trùng tên file: Nếu file đã tồn tại ở đích, đổi tên file mới
            if os.path.exists(dest_path):
                name, ext = os.path.splitext(filename)
                dest_path = os.path.join(destination, f"{name}_from_other_folder{ext}")

            # Thực hiện copy
            shutil.copy2(source_path, dest_path)
            total_copied += 1

    print(f"\n" + "="*30)
    print(f"HOÀN THÀNH: Đã gộp tổng cộng {total_copied} ảnh vào {destination}")
    print("="*30)

# Chạy hàm gộp
merge_folders([folder_1, folder_2,folder_3], target_folder)
```

```python
import os
import shutil
from google.colab import drive

# 1. Kết nối Google Drive
drive.mount('/content/drive')

# 2. Cấu hình đường dẫn
folder_1 = '/content/drive/MyDrive/data/House/'
folder_2 = '/content/drive/MyDrive/data/Transport/'
folder_3 = '/content/drive/MyDrive/data/Street/'
folder_4 = '/content/drive/MyDrive/data/bo_sung/'
target_folder = '/content/drive/MyDrive/data/8000 ảnh'

# Tạo folder đích nếu chưa có
if not os.path.exists(target_folder):
    os.makedirs(target_folder)
    print(f"Đã tạo folder mới: {target_folder}")

def merge_and_count(source_folders, destination):
    valid_exts = ('.jpg', '.jpeg', '.png', '.bmp', '.gif')
    summary = {} # Dùng để lưu số lượng ảnh của từng folder
    total_copied = 0

    print("--- BẮT ĐẦU THỐNG KÊ VÀ GỘP ẢNH ---")

    for src in source_folders:
        # Kiểm tra nếu folder tồn tại
        if os.path.exists(src):
            # Lấy danh sách ảnh trong folder hiện tại
            files = [f for f in os.listdir(src) if f.lower().endswith(valid_exts)]
            count_in_src = len(files)

            # Lưu vào báo cáo
            folder_name = os.path.basename(os.path.normpath(src))
            summary[folder_name] = count_in_src

            print(f"\nFolder [{folder_name}]: Tìm thấy {count_in_src} ảnh. Đang copy...")

            for filename in files:
                source_path = os.path.join(src, filename)
                dest_path = os.path.join(destination, filename)

                # Xử lý trùng tên
                if os.path.exists(dest_path):
                    name, ext = os.path.splitext(filename)
                    dest_path = os.path.join(destination, f"{name}_from_{folder_name}{ext}")

                shutil.copy2(source_path, dest_path)
                total_copied += 1
        else:
            print(f"\nCảnh báo: Không tìm thấy folder {src}")

    # Đếm tổng số file thực tế trong folder đích sau khi gộp
    final_count = len([f for f in os.listdir(destination) if f.lower().endswith(valid_exts)])

    # --- IN BÁO CÁO CHI TIẾT ---
    print("\n" + "="*40)
    print("BÁO CÁO SỐ LƯỢNG ẢNH")
    print("="*40)
    for folder, count in summary.items():
        print(f" - {folder.ljust(15)}: {count} ảnh")

    print("-" * 40)
    print(f"Tổng số ảnh đã copy thành công: {total_copied}")
    print(f"Tổng số ảnh hiện có tại đích  : {final_count}")
    print("="*40)

# Chạy hàm
f = [folder_1,folder_2,folder_3,folder_4]
merge_and_count(f, target_folder)
```

```python
import os
import re
from google.colab import drive

# 1. Kết nối Drive
if not os.path.exists('/content/drive'):
    drive.mount('/content/drive')

# --- CẤU HÌNH ĐƯỜNG DẪN THƯ MỤC ẢNH ---
image_folder_path = '/content/drive/MyDrive/data/8000 ảnh'

def extract_number(filename):
    """Trích xuất số từ tên file để sắp xếp (00005.png -> 5)"""
    nums = re.findall(r'\d+', filename)
    return int(nums[0]) if nums else 0

def check_and_sort_non_jpg(folder_path):
    if not os.path.exists(folder_path):
        print(f"❌ Không tìm thấy thư mục: {folder_path}")
        return

    all_files = os.listdir(folder_path)
    non_jpg_files = []

    print(f"📁 Tổng số file trong thư mục: {len(all_files)}")

    for filename in all_files:
        # Bỏ qua thư mục con
        if os.path.isdir(os.path.join(folder_path, filename)):
            continue

        # Kiểm tra đuôi file (không phân biệt hoa thường)
        _, ext = os.path.splitext(filename)
        if ext.lower() != '.jpg':
            non_jpg_files.append(filename)

    if not non_jpg_files:
        print("✅ Tuyệt vời! Tất cả các file đều là định dạng .jpg")
        return

    # Sắp xếp các file không phải .jpg theo thứ tự số
    non_jpg_files.sort(key=extract_number)

    print("-" * 50)
    print(f"⚠️ Tìm thấy {len(non_jpg_files)} file KHÔNG PHẢI .jpg")
    print("-" * 50)

    # Hiển thị 10 file đầu tiên
    num_to_show = 10
    print(f"📋 {min(num_to_show, len(non_jpg_files))} file ĐẦU TIÊN trong danh sách lỗi:")
    for f in non_jpg_files[:num_to_show]:
        print(f"  - {f}")

    print("\n...")

    # Hiển thị 10 file cuối cùng
    if len(non_jpg_files) > num_to_show:
        print(f"📋 {min(num_to_show, len(non_jpg_files))} file CUỐI CÙNG trong danh sách lỗi:")
        for f in non_jpg_files[-num_to_show:]:
            print(f"  - {f}")

    # Thống kê nhanh các loại đuôi file lạ
    extensions = [os.path.splitext(f)[1].lower() for f in non_jpg_files]
    from collections import Counter
    print("\n📊 Thống kê các định dạng lạ:")
    for ext, count in Counter(extensions).items():
        print(f"  - {ext if ext else '[Không đuôi]'}: {count} file")

# Chạy kiểm tra
check_and_sort_non_jpg(image_folder_path)
```

rename sang jpg

```python
import os
import re
from google.colab import drive

# 1. Kết nối Google Drive
if not os.path.exists('/content/drive'):
    drive.mount('/content/drive')

# --- CẤU HÌNH ĐƯỜNG DẪN ---
# Thay đường dẫn này bằng thư mục chứa ảnh của bạn
image_folder_path = '/content/drive/MyDrive/data/8000 ảnh'

def rename_to_jpg_standard():
    if not os.path.exists(image_folder_path):
        print(f"❌ Không tìm thấy thư mục: {image_folder_path}")
        return

    files = os.listdir(image_folder_path)
    print(f"🚀 Tổng số file phát hiện: {len(files)}")
    print("⏳ Đang bắt đầu quá trình đổi tên và định dạng...")

    renamed_count = 0
    skipped_count = 0
    error_count = 0

    for filename in files:
        old_path = os.path.join(image_folder_path, filename)

        # Bỏ qua nếu là thư mục con
        if os.path.isdir(old_path):
            continue

        # 1. Lấy số thứ tự từ tên file (ví dụ: '01276' từ '1276.png' hoặc '01276.jpeg')
        numbers = re.findall(r'\d+', filename)

        if numbers:
            # Chuyển số đầu tiên tìm thấy thành định dạng 0000x.jpg
            num = int(numbers[0])
            new_filename = f"{num:05d}.jpg"
            new_path = os.path.join(image_folder_path, new_filename)

            # 2. Thực hiện đổi tên nếu tên mới khác tên cũ
            if filename != new_filename:
                try:
                    # Nếu file đích đã tồn tại (tránh ghi đè file cũ)
                    if os.path.exists(new_path):
                        print(f"⚠️  Bỏ qua {filename}: Đã có file {new_filename} tồn tại.")
                        skipped_count += 1
                    else:
                        os.rename(old_path, new_path)
                        renamed_count += 1
                except Exception as e:
                    print(f"❌ Lỗi khi xử lý {filename}: {e}")
                    error_count += 1
            else:
                # File đã đúng định dạng rồi
                skipped_count += 1
        else:
            # Những file không chứa số (như file ẩn .DS_Store) thì bỏ qua
            # print(f"ℹ️  Bỏ qua file không chứa số: {filename}")
            skipped_count += 1

    print("-" * 50)
    print(f"✅ HOÀN TẤT QUÁ TRÌNH!")
    print(f"📸 Số ảnh đã đổi sang .jpg chuẩn: {renamed_count}")
    print(f"🆗 Số ảnh đã chuẩn sẵn hoặc bỏ qua: {skipped_count}")
    if error_count > 0:
        print(f"⚠️ Số lỗi phát sinh:               {error_count}")
    print("-" * 50)

# Chạy code
rename_to_jpg_standard()
```

Resize ảnh xuống 224x224

```python
import os
from PIL import Image
from tqdm import tqdm # Thư viện để hiển thị thanh tiến trình
from google.colab import drive

# 1. Kết nối Drive
if not os.path.exists('/content/drive'):
    drive.mount('/content/drive')

# --- CẤU HÌNH ĐƯỜNG DẪN ---
input_folder = '/content/drive/MyDrive/data/8000 ảnh'       # Folder ảnh gốc (đã chuẩn hóa .jpg)
output_folder = '/content/drive/MyDrive/data/images'  # Folder lưu ảnh sau khi resize
target_size = (224, 224)

def resize_images():
    # Tạo folder đầu ra nếu chưa có
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
        print(f"📁 Đã tạo thư mục mới: {output_folder}")

    # Lấy danh sách file ảnh
    files = [f for f in os.listdir(input_folder) if f.lower().endswith('.jpg')]
    print(f"🚀 Tìm thấy {len(files)} ảnh. Bắt đầu resize về {target_size}...")

    success_count = 0
    error_count = 0

    # Sử dụng tqdm để theo dõi tiến độ
    for filename in tqdm(files, desc="Processing"):
        try:
            img_path = os.path.join(input_folder, filename)
            save_path = os.path.join(output_folder, filename)

            # Mở ảnh
            with Image.open(img_path) as img:
                # Chuyển về hệ màu RGB (tránh lỗi khi gặp ảnh PNG/RGBA cũ)
                img = img.convert('RGB')

                # Resize ảnh sử dụng thuật toán LANCZOS (cho chất lượng tốt nhất)
                img_resized = img.resize(target_size, Image.Resampling.LANCZOS)

                # Lưu ảnh với chất lượng cao (95)
                img_resized.save(save_path, "JPEG", quality=95)

            success_count += 1
        except Exception as e:
            print(f"\n❌ Lỗi khi xử lý file {filename}: {e}")
            error_count += 1

    print("-" * 50)
    print(f"✅ HOÀN TẤT!")
    print(f"🖼️  Đã resize thành công: {success_count} ảnh.")
    if error_count > 0:
        print(f"⚠️  Số ảnh bị lỗi:        {error_count} ảnh.")
    print(f"📂 Ảnh mới nằm tại: {output_folder}")
    print("-" * 50)

# Chạy thực hiện
resize_images()
```
